# LIAR DATASET

Link [LIAR Dataset](https://www.kaggle.com/datasets/luks06/liar-dataset)

```python
import pandas as pd
```

```python
liar_dataset_test = pd.read_csv("../data/liar_dataset/test.tsv", sep="\t", header=None)
liar_dataset_train = pd.read_csv("../data/liar_dataset/train.tsv", sep="\t", header=None)
liar_dataset_valid = pd.read_csv("../data/liar_dataset/valid.tsv", sep="\t", header=None)
```

```python

columns_names = ["id", "label_text", "text", "subject", "speaker", "job_title", "state_info", "party_affiliation", "barely_true_counts", "false_counts", "half_true_counts", "mostly_true_counts", "pants_on_fire_counts", "context"]
liar_dataset_test.columns = columns_names
liar_dataset_train.columns = columns_names
liar_dataset_valid.columns = columns_names
liar_dataset = pd.concat([liar_dataset_test, liar_dataset_train, liar_dataset_valid], ignore_index=True)
```

```python
liar_dataset = liar_dataset.copy()
liar_dataset = liar_dataset[["text", "label_text"]]
liar_dataset["label_text"].value_counts()
```

```python
condition = (liar_dataset["label_text"] != "half-true") & (liar_dataset["label_text"] != "barely-true")
liar_dataset = liar_dataset[condition]
liar_dataset["label_text"].value_counts()
```

```python
condition_true = (liar_dataset["label_text"] == "true") | (liar_dataset["label_text"] == "mostly-true")
liar_dataset["label"] = condition_true.astype("int32")
liar_dataset = liar_dataset.drop(columns=["label_text"])
```

### Real News Pro

https://www.kaggle.com/datasets/yousefzahran1/realnews-pro

```python
real_news_pro_df = pd.read_csv("../data/RealNews-Pro/combined_fake_news_3.csv")
```

```python
real_news_pro_df = real_news_pro_df[["text", "label"]].reset_index(drop=True)
real_news_pro_df["label"] = real_news_pro_df["label"].astype(str)
real_news_pro_df = real_news_pro_df[~real_news_pro_df["text"].str.contains(r'\[\+\d+\s*chars\]')]
real_news_pro_df.to_csv("../data/datasets/real_news_pro_dataset.csv")
```

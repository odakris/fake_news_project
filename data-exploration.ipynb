### fake-and-real-news-dataset

```python
FAKE_fake_and_real_news_dataset = pd.read_csv('./data/fake-and-real-news-dataset/Fake.csv')
TRUE_fake_and_real_news_dataset = pd.read_csv('./data/fake-and-real-news-dataset/True.csv')
```

```python
fake_subject = FAKE_fake_and_real_news_dataset['subject'].value_counts()
true_subject = TRUE_fake_and_real_news_dataset['subject'].value_counts()
print("Fake news subjects:" + str(fake_subject))
print("True news subjects:" + str(true_subject))
```

```python
message = FAKE_fake_and_real_news_dataset['text'][0]
message
```

```python
message = message.lower()
message = re.sub(r'|http\S+|www\.\S+|@\w+|[^\w\s]', '', message)
message
```

```python
FAKE_fake_and_real_news_dataset.isnull().sum()
```

```python
TRUE_fake_and_real_news_dataset.isnull().sum()
```

```python
print("Fake text nulls:", FAKE_fake_and_real_news_dataset['text'].isnull().sum())
print("True text nulls:", TRUE_fake_and_real_news_dataset['text'].isnull().sum())
print("Fake text empty strings:", (FAKE_fake_and_real_news_dataset['text'] == '').sum())
print("True text empty strings:", (TRUE_fake_and_real_news_dataset['text'] == '').sum())
print("Fake text whitespace-only:", FAKE_fake_and_real_news_dataset['text'].str.strip().eq('').sum())
print("True text whitespace-only:", TRUE_fake_and_real_news_dataset['text'].str.strip().eq('').sum())
print("Total whitespace-only:", FAKE_fake_and_real_news_dataset['text'].str.strip().eq('').sum() + TRUE_fake_and_real_news_dataset['text'].str.strip().eq('').sum())
```
